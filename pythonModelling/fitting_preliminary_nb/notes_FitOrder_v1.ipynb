### This is the full order of the fitting for the VUV4 data.
PDE equation is as below. Don't ask you should know
$$ PDE = FF \cdot T(\lambda, \theta_1, t_{oxide}) \cdot \big( W_p(\lambda, dp^*, X_{PN}) p_e(V) + W_n(\lambda, X_{PN}, dw^*) p_h(V)\big)$$

In the VUV (below ~3.7eV or 335nm) $p_e$ changes to accomodate the quantum yield term which is expected to be greater than 1. $N_Y(\lambda)$ is the average number of electron/hole pairs made$. Maybe treating it as binomial with average $N_Y$ is different mathematically but should be close enough.
$$p_e \rightarrow 1 - (1-p_e)^{N_Y}$$
