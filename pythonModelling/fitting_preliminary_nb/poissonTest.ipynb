```python
import numpy as np
import matplotlib.pyplot as plt

# Example usage
lam_laser = 0.8 # poisson rate for laser
lamCA_pure = 0.3 # poisson rate for CA, for a single pulse

Nrun = 1000000 

NP_laser = np.random.poisson(lam_laser, Nrun) # number of pulses from laser
NP_ca = []

for Npulse in NP_laser:
    lambdaCA = lamCA_pure*Npulse # this multiplication by Npulse is what changes
    # the final, calculated CA average to be not equal to lamCA_pure.
    
    Nextra = np.random.poisson(lambdaCA)
    NP_ca.append(Nextra)
    
    # print(Npulse, Nextra)

NP_ca = np.array(NP_ca)
NP_measured = NP_laser + NP_ca

```

```python
plt.hist(NP_laser, range=(-0.5, 9.5), bins=10, alpha=0.5, label='laser')
plt.hist(NP_ca, range=(-0.5, 9.5), bins=10, alpha=0.5, label='CA')
plt.hist(NP_measured, range=(-0.5, 9.5), bins=10, alpha=0.5, label='measured')

plt.legend()
plt.show()
```

```python
mLaser, mCA, mData = np.mean(NP_laser), np.mean(NP_ca), np.mean(NP_measured)
# print(mLaser, mCA, mData)

# calculated_mCA = mData - mLaser 
# print(calculated_mCA)
# print(mCA)

print("laser data mean:", mLaser)
# print("(hidden) CA mean:", mCA)
print("data mean:", mData)

# inputAverage = lam_laser*(1+lamCA_pure)
# print("input average:", inputAverage)
# l3 = l1 x (1+l2)

mCa_fromData = mData/mLaser - 1
print("measured CA:", mCa_fromData)
```
