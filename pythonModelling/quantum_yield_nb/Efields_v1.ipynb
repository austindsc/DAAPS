```python
import numpy as np

def E_p(x, V, Ln, Lp):
    return 2 * V / (Ln + Lp) * (x/Lp + 1)

def E_n(x, V, Ln, Lp):
    return 2*V/(Ln + Lp) * (1 - x/Ln)
```

```python
import matplotlib.pyplot as plt

Lp = 2
Ln = 5

xP = np.linspace(-Lp, 0, 100)
xN = np.linspace(0, Ln, 100)

V = [1, 5, 10]
for v in V:
    plt.plot(xP, E_p(xP, v, Ln, Lp), label=f'E_p, V={v}')
    plt.plot(xN, E_n(xN, v, Ln, Lp), label=f'E_n, V={v}')
    
pIntegral = np.sum(E_p(xP, 1, Ln, Lp)) * (xP[1] - xP[0])
nIntegral = np.sum(E_n(xN, 1, Ln, Lp)) * (xN[1] - xN[0])

print(f'pIntegral = {pIntegral}')
print(f'nIntegral = {nIntegral}')
```
